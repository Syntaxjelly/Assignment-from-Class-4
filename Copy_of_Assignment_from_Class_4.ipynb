```python
#mathemetical opperation on array
a = np.array([1, 2, 3, 4, 5])
b = np.array([2, 3, 4, 5, 6])
c=np.array([4,5,6,5,6])
# Element-wise operations
print("Addition:",a+b+c)
print("Multiplycation:",(a*b)*c)
print("power:",b**3)
print("Squire root:",np.sqrt(c))

# Statistical operations
print("Mean:",np.mean(a))
print("standart deviation:",np.std(a))
print("min:",np.min(a))
print("max:",np.max(c))

#Broadcasting array
matrix=np.array([[1,2,3],[4,5,6]])
vector=np.array([10,20,30])
result=matrix+vector
print("Broadcasting array: \n",result)
```

# ***Excercise for numpy ***

```python
# Import necessary library
import numpy as np

# Create a NumPy array with 20 random integers between 1 and 100
random_array=np.random.randint(1,101,20)

# Display the NumPy array
print("Random number: \n",random_array)
# 1. Find the sum of all elements
total_sum=np.sum(random_array)
print("total sum:",total_sum)


# 2. Find the maximum and minimum values in the array
max_value = np.max(random_array)
print("Maximumu value of array:",max_value)
min_value=np.min(random_array)
print("Minimum value of array:",min_value)
```
